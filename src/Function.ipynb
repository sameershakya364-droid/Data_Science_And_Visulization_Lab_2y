```python
#function argument
#you can call a function by using the following types of formal argument
# .default argument
# .required argument
# .keyword argument
# .variable-length
```

```python
#Default Argument
#function define is here
def printinfo(name,age = 35):
  print ("Name: " , name)
  print ("Age: " , age)
  return;
```

```python
#now you can call printinfo function
printinfo(age=50,name="Sam")
printinfo(name="Sam", age=35)
```

```python
#required argument
def printinfo(name,age):
  print ("Name: " , name)
  print ("Age: " , age)
  return;
```

```python
printinfo("miki", 30)
```

```python
#keyword argument
#function definition is here
def printme(string_to_print, rollno):
  """this prints a passed string into this function"""
  print(string_to_print)
  return
```

```python
#now you can call printme function
printme(string_to_print="how are you", rollno=21)
```

```python
#function definition is here
def printinfo(arg1, *vartuple):
  print("Output is: ")
  print(arg1)
  for var in vartuple:
    print(var)
  return;
```

```python
#now you can call prininfo function
printinfo(10)
printinfo(70,60,50)
```

```python
#variable-length argument
#function definition is here
def printinfo(arg1, *vartuple):
  print("Output is: ")
  print(arg1)
  for var in vartuple:
    print(var)
  return;
```

```python
#now you can call printinfo function
printinfo(10)
```

```python
printinfo(70,60,50)
```
