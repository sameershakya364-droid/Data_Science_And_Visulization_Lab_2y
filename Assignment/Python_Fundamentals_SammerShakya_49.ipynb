## Topic 3: Objects, Identity (is) vs Equality (==)

```python
# Q3.1
# == checks whether two values are equal.
# is checks whether two variables refer to the same object in memory.
```

```python
# Q3.2
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```

```python
# Q3.3
x = 100
y = 100

print(x is y)
print(x == y)

p = 1000
q = 1000

print(p is q)
print(p == q)

# Python may reuse small integer objects, so small integers can have
# the same identity. For larger integers, identity can differ even
# when their values are equal. Therefore == should be used for values.
```

## Topic 7: Functions & Passing Arguments

```python
# Q7.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

```python
# Q7.2
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

# my_abs("Hi")
# This causes TypeError because '<' is not supported between str and int.

try:
    print(my_abs("Hi"))
except TypeError as e:
    print("TypeError:", e)
```

```python
# Q7.3
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)

# 2.7 is printed first.
# x is None because the function prints the value but does not return it.
```

```python
# Q7.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)

# Output: 7
# Reassigning val only changes the local variable inside the function.
```

```python
# Q7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:", val1, val2)

x = 6
y = 3

swap_message(x, y)
print("Original variables:", x, y)
```

## Topic 8: Scope — Local vs Global Variables

```python
# Q8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

try:
    print(val)
except NameError as e:
    print("NameError:", e)

# val is local to my_abs(), so it cannot be accessed outside the function.
```

```python
# Q8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3
counter = 0

def increment_counter():
    # Without global, Python treats counter as a local variable
    # because it is assigned inside this function.
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
```

```python
# Q8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
